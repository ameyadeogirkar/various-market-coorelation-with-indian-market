```python
# Install libraries (Colab)
!pip install yfinance --quiet

import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# Tickers
tickers = {
    "NIFTY 50": "^NSEI",
    "Dow Futures": "YM=F",
    "Nasdaq Futures": "NQ=F"
}

# Download 5 years of data
data = {}

for name, ticker in tickers.items():
    df = yf.download(
        ticker,
        period="5y",
        auto_adjust=True,
        progress=False
    )
    data[name] = df

# Create side-by-side charts
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, (name, df) in zip(axes, data.items()):
    ax.plot(df.index, df["Close"], linewidth=1.5)
    ax.set_title(name)
    ax.set_xlabel("Date")
    ax.set_ylabel("Close Price")
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Normalized comparison (base = 100)

fig, ax = plt.subplots(figsize=(12, 6))

for name, df in data.items():
    normalized = (df["Close"] / df["Close"].iloc[0]) * 100
    ax.plot(normalized.index, normalized, label=name)

ax.set_title("5-Year Performance Comparison (Base = 100)")
ax.set_xlabel("Date")
ax.set_ylabel("Indexed Value")
ax.legend()
ax.grid(True, alpha=0.3)

plt.show()
```
